```python
my_dict = {'a': 1, 'b': 2, 'c': 3, 'd': 4}
keys_to_remove = ['b', 'd']

filtered_dict = {k: v for k, v in my_dict.items() if k not in keys_to_remove}
print(filtered_dict)  # Output: {'a': 1, 'c': 3}
```

```python
my_dict = {'a': 1, 'b': 2, 'c': 1, 'd': 3}
value_to_remove = 1

filtered_dict = {k: v for k, v in my_dict.items() if v != value_to_remove}
print(filtered_dict)  # Output: {'b': 2, 'd': 3}
```

```python
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}

# Using update()
dict1.update(dict2)
print(dict1)  # Output: {'a': 1, 'b': 2, 'c': 3, 'd': 4}

# Using dictionary unpacking
merged_dict = {**dict1, **dict2}
print(merged_dict)  # Output: {'a': 1, 'b': 2, 'c': 3, 'd': 4}
```

```python
my_dict = { 'b': 2, 'a': 1, 'c': 3}
sorted_dict = dict(sorted(my_dict.items()))
print(sorted_dict)  # Output: {'a': 1, 'b': 2, 'c': 3}
```

```python
sorted_by_value = dict(sorted(my_dict.items(), key=lambda item: item[1]))
print(sorted_by_value)  # Output: {'a': 1, 'b': 2, 'c': 3}
```

```python
def check_even_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"
# Test the function   

print(check_even_odd(10))  # Output: Even
print(check_even_odd(15))  # Output: Odd
```

```python
def check_num(num):
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"
# Test the function
print(check_num(10))  # Output: Positive
print(check_num(-5))  # Output: Negative
print(check_num(0))   # Output: Zero
```

```python
def palindrome(s):
    return s == s[::-1]
# Test the function
print(palindrome("racecar"))  # Output: True
print(palindrome("hello"))    # Output: False
```

```python
def remove_duplicates(lst):
    return list(set(lst))
# Test the function
print(remove_duplicates([1, 2, 2, 3, 4, 4]))  # Output: [1, 2, 3, 4]
```

```python
def remove_duplicates(lst):
    return list(dict.fromkeys(lst)) # Using dict.fromkeys() to remove duplicates
# Test the function
print(remove_duplicates([1, 2, 2, 3, 4, 4]))  # Output: [1, 2, 3, 4]
```
